```python
## Dynamic Typing
## Python allows type of variable to change as the program executes

var = 10 # int
print(var,type(var))

var = "Ashitosh" # str
print(var,type(var))

var = 3.14 # float
print(var,type(var))
```

```python
### Simple Calculator

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:",sum)
print("Difference:",difference)
print("Product:",product)
print("Quotient:",quotient)
```
