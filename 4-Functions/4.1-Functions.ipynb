```python
## Why function ?

num = 24
if num % 2 == 0:
    print("The number is Even")
else:
    print("The number is Odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num % 2 == 0:
        print("The number is Even")
    else:
        print("The number is Odd")
```

```python
## Call this function
even_or_odd(24)
```

```python
## Function with multiple parameters

def add(a, b):
    return a + b

## Call this function
result = add(10,11)
print(result)
```

```python
## Default parameters

def greet(name):
    print(f"Hello {name}")

greet("Ashitosh")
```

```python
def greet(name = "Ashitosh"):
    print(f"Hello {name}")

greet()
```

```python
## Variable Length Arguments
## Positional And Keyword Arguments

def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6)
```

```python
## Keyword Arguments

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")
        
```

```python
print_details(name = "Ashitosh", age = 21, Country = "India")
```

```python
def print_details(*args, **kwargs):
    for val in args:
        print(f"Postional arguments:{val}")
    for key, value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(1,2,3,4,5,6,name = "Ashitosh", age = 21, Country = "India")
```

```python
## Return Statement

def multiply(a, b):
    return a * b

multiply(2, 3)
```

```python
## Return multiple parameters

def multiply(a, b):
    return a * b, a

multiply(2, 3)
```
