```python
## Syntax

    lambda arguments: expression
```

```python
def addition(a, b):
    return a + b

addition(2,3)
```

```python
addition = lambda a, b: a + b

print(type(addition))
print(addition(5, 6))
```

```python
def even(n):
    if n % 2 == 0:
        return True

even(12)
```

```python
even1 = lambda num: num%2 == 0

even1(12)
```

```python
def addition(x, y, z):
    return x + y + z

addition(12,11,14)
```

```python
addition = lambda x, y , z : x + y + z

addition(12,11,14)
```

```python
## map() - applies a function to all items in a list
number = [1,2,3,4,5,6]
def square(number):
    return number**2

square(2)
```

```python
list(map(lambda x: x**2,number))
```
