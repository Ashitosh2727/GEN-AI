```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
paragraph = """Natural language processing is an exciting field of artificial intelligence that focuses on the interaction between humans and computers using natural language. It plays a crucial role in many modern applications, including chatbots, search engines, sentiment analysis, machine translation, and virtual assistants. When we process textual data, one of the very first steps is cleaning and preprocessing the data to make it suitable for analysis. This involves removing unnecessary characters, punctuation marks, and common words that do not carry significant meaning, known as stopwords. Stopwords include words like "is," "and," "the," "in," "of," and "on." These words occur frequently in sentences but usually do not add much value to the overall meaning of the text. For example, in the sentence, "The quick brown fox jumps over the lazy dog," the words "the," "over," and "on" are stopwords. By removing such words, we can reduce the dataset size and focus on more meaningful words such as "quick," "brown," "fox," "jumps," and "lazy." This step improves the efficiency of algorithms and models, especially when working with large volumes of data. However, it is essential to note that stopword removal must be handled carefully because, in certain contexts, these words can affect the interpretation of a sentence. For instance, in sentiment analysis, the word "not" is very important, and removing it could completely change the meaning of a statement like "I do not like this product." Therefore, customizing the stopword list based on the specific task and domain is a crucial part of building robust NLP pipelines."""
```

```python
stemmer = PorterStemmer()
```

```python
sentences = nltk.sent_tokenize(paragraph)
```

```python
type(sentences)
```

```python
## Apply Stopwords And Filter And then Apply Stemming

for i in range (len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # Converting all the list of words in sentences   
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
SnowballsStemmer = SnowballStemmer('english')
```

```python
## Apply Stopwords And Filter And then Apply Snowball Stemmer

for i in range (len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [SnowballsStemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # Converting all the list of words in sentences   
```

```python
sentences
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4') 
lemmatizer = WordNetLemmatizer()
```

```python
## Apply Stopwords And Filter And then Apply lammatizer

for i in range (len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word.lower()) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words) # Converting all the list of words in sentences   
```
