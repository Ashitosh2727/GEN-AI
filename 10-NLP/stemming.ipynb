```python
import nltk
from nltk.stem import PorterStemmer
```

```python
## craeting an object
porter = PorterStemmer()
```

```python
## Classifiaction Problem
## Comments of product is a positive review or negative review
## Reviews ---> [eating, eat, eaten] -> eat [going, gone, goes] --> go

words = ["eating", "eats", "eaten", "writing", "writes", "programming", "programs", "history", "finally", "finalized"]
```

```python
for word in words:
    print(word+"---->"+porter.stem(word))
```

```python
print(porter.stem("congratulations"))
print(porter.stem("sitting"))
```

### RegexpStemmer

```python
from nltk.stem import RegexpStemmer
```

```python
## creating an object
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$')
```

```python
reg_stemmer.stem("eating")
```

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
SnowballsStemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+SnowballsStemmer.stem(word))
```

```python
print(porter.stem("fairly"))
print(porter.stem("sportingly"))
```

```python
print(SnowballsStemmer.stem("fairly"))
print(SnowballsStemmer.stem("sportingly"))
```
