```python
from abc import ABC, abstractmethod

# Abstract base class
class Vehicle(ABC):
    def drive(self):
        print("Driving the vehicle")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Started the Car engine")
    
def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car = Car()
operate_vehicle(car)
```
