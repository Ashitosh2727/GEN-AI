```python
## Creating Dictionaries
empty_dict = {}
print(type(empty_dict))
```

```python
student = {"name": "Ashitosh","age" : 21, "grade" : 24}
print(student)
print(type(student))
```

```python
## Error
student = {"name": "Ashitosh","age" : 21, "name" : 24}
print(student)
```

```python
## Accessing Dictionary elements
student = {"name": "Ashitosh","age" : 21, "grade" : 24}

print(student['name'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
```

```python
## Modifying Dictionary elements
## Dictionary are mutable , so you can add,update or delete elements

print(student)
```

```python
student['age'] = 22 ## Update the for the key
print(student)

student['address'] = "India" ## Add new key-value pair
print(student)
```

```python
del student['address'] ## Delete key-value pair
print(student)
```

```python
## Dictionary Methods

keys = student.keys()  ## Get all keys
print(keys)

values = student.values()  ## Get all values
print(values)

items = student.items()  ## Get all key-value pairs
print(items)
```

```python
## Shallow Copy
student_copy = student
print(student)
print(student_copy)
```

```python
student['name'] = "Ashitosh Mane"
print(student)
print(student_copy)  ## Both will change as they point to the same object
```

```python
student_copy1 = student.copy()  ## Create a shallow copy
print(student_copy1)
print(student)
```

```python
student['name'] = "Ashitosh Mohan Mane"
print(student_copy1)  
print(student)
```

```python
## Itearating over a Dictionary

student = {"name": "Ashitosh","age" : 21, "grade" : 24}

## iterating over keys
for keys in student.keys():
    print(keys)
    
```

```python
## iterating over values
for values in student.values():
    print(values)
```

```python
## iterating over key-value pairs

for key, value in student.items():
    print(f"{key} : {value}")
```

```python
## Nested Dictionaries
student = {
    "student1" : {"Name" : "Ashitosh", "Age" : 21 },
    "student2" : {"Name" : "John", "Age" : 22 },
}
print(student)
```

```python
## Accessing nested dictionary
print(student["student2"]["Name"])
print(student["student2"]["Age"])
```

```python
## iterating over nested dictionaries
for student_id ,student_info in student.items():
    print(f"{student_id} : {student_info}")
    for key , value in student_info.items():
        print(f"{key} : {value}")
```

```python
## Dictionary Comprehension
squares = {x: x**2 for x in range(5)}
print(squares)
```

```python
## Conditional Dictionary Comprehension
evens = {x: x**2 for x in range(10) if x % 2 == 0}
print(evens)
```

```python
## Practical Examples

## Use a dictionary to count the frequency of each element in a list

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print(frequency)
```

```python
## Merge two dictionaries in one

dict1 = {'a':1,'b': 2}
dict2 = {'b':3,'c': 4}
merge_dict = {**dict1, **dict2}
print(merge_dict)
```
