```python
## Packing and Unpacking Tuple
## Packing
packed_tuple = 1,"Hello",3.14
print(packed_tuple)
```

```python
## Unpacking a Tuple
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking With *

numbers = (1,2,3,4,5,6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```
